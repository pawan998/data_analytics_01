### Step 04: EDA (Exploratery Data Analysis)

```python
# Create a bar chart showing data distribution by state
state_counts = loan_df['state'].value_counts().sort_values(ascending=False)

plt.figure(figsize=(14, 6))
sns.barplot(data=pd.DataFrame({'state': state_counts.index, 'count': state_counts.values}), 
            x='state', y='count', hue='state', palette='viridis', legend=False)
plt.title('Data Distribution by State', fontsize=16, fontweight='bold')
plt.xlabel('State', fontsize=12)
plt.ylabel('Count of Records', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Display summary statistics
print(f"\nTotal Records: {state_counts.sum()}")
print(f"\nTop 10 States by Record Count:")
print(state_counts.head(10))
```
